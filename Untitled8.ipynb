# Student Performance Prediction and Model Evaluation

This notebook will guide you through a complete machine learning workflow on the Student Performance dataset, covering Exploratory Data Analysis (EDA), regression modeling, and classification modeling.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
try:
    df = pd.read_csv('student_performance.csv')
    print("Dataset 'student_performance.csv' loaded successfully!")
except FileNotFoundError:
    print("Error: 'student_performance.csv' not found. Please upload the dataset or specify the correct path.")
    df = None # Set df to None if loading fails

if df is not None:
    # Display the first 5 rows of the dataset
    print("\nFirst 5 rows of the dataset:")
    display(df.head())

    # Display basic information about the dataset (columns, data types, non-null values)
    print("\nDataset Information:")
    df.info()

    # Display descriptive statistics for numerical columns
    print("\nDescriptive Statistics:")
    display(df.describe())
```

## Exploratory Data Analysis (EDA)

In this section, we will explore the dataset to understand its characteristics, identify numerical and categorical variables, examine distributions, relationships, correlations, and check for missing values and potential outliers. We will also try to determine factors related to the `exam_score`.

```python
if df is not None:
    # Identify numerical and categorical variables
    numerical_cols = df.select_dtypes(include=np.number).columns.tolist()
    categorical_cols = df.select_dtypes(include='object').columns.tolist()

    print(f"\nNumerical Columns: {numerical_cols}")
    print(f"Categorical Columns: {categorical_cols}")

    # Check for missing values
    print("\nMissing Values per Column:")
    display(df.isnull().sum().sort_values(ascending=False))

    # Visualize distributions of numerical features
    print("\nDistributions of Numerical Features:")
    plt.figure(figsize=(15, 10))
    for i, col in enumerate(numerical_cols):
        plt.subplot(len(numerical_cols) // 3 + 1, 3, i + 1)
        sns.histplot(df[col], kde=True)
        plt.title(f'Distribution of {col}')
        plt.tight_layout()
    plt.show()
else:
    print("Cannot perform EDA as the DataFrame is not loaded.")
```

```python
if df is not None:
    # Visualize distributions of categorical features
    print("\nDistributions of Categorical Features:")
    plt.figure(figsize=(15, 10))
    for i, col in enumerate(categorical_cols):
        plt.subplot(len(categorical_cols) // 3 + 1, 3, i + 1)
        sns.countplot(y=df[col], order=df[col].value_counts().index, palette='viridis')
        plt.title(f'Count of {col}')
        plt.tight_layout()
    plt.show()
else:
    print("Cannot perform EDA as the DataFrame is not loaded.")
```
