## Food Delivery Business Performance Analysis

This notebook will explore the provided food delivery business performance dataset to uncover meaningful patterns and insights through data visualization.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
# Please ensure your CSV file is uploaded to your Colab environment or specify the correct path.
try:
    df = pd.read_csv('food_delivery_data.csv')
    print("Dataset loaded successfully!")
except FileNotFoundError:
    print("Error: 'food_delivery_data.csv' not found. Please upload the dataset or update the file path.")
    df = None # Set df to None if loading fails

if df is not None:
    # Display the first 5 rows of the dataset
    print("\nFirst 5 rows of the dataset:")
    display(df.head())

    # Display basic information about the dataset (columns, data types, non-null values)
    print("\nDataset Information:")
    df.info()
```

### Data Cleaning and Preparation

Before visualization, we'll perform some basic cleaning steps, such as converting date columns to datetime objects.

```python
if df is not None:
    # Attempt to convert a common date column name to datetime objects
    # You might need to adjust 'order_date' to the actual date column in your dataset
    date_column_candidates = ['order_date', 'date', 'delivery_date', 'timestamp']
    for col in date_column_candidates:
        if col in df.columns:
            try:
                df[col] = pd.to_datetime(df[col])
                print(f"Column '{col}' converted to datetime.")
                # Let's assume 'order_date' is the primary date column for time series analysis
                df = df.sort_values(by=col).reset_index(drop=True)
                break
            except Exception as e:
                print(f"Could not convert '{col}' to datetime: {e}")

    # Display updated info to check date type conversion
    print("\nUpdated Dataset Information:")
    df.info()
```

### Visualization Portfolio

Let's start by visualizing trends over time, such as changes in orders and revenue.

```python
if df is not None and 'order_date' in df.columns and 'order_id' in df.columns:
    # Aggregate daily orders
    daily_orders = df.groupby('order_date')['order_id'].nunique().reset_index()
    daily_orders.columns = ['Date', 'Number of Orders']

    plt.figure(figsize=(12, 6))
    sns.lineplot(x='Date', y='Number of Orders', data=daily_orders)
    plt.title('Daily Number of Orders Over Time')
    plt.xlabel('Date')
    plt.ylabel('Number of Orders')
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

    print("\nInterpretation: This line plot shows the trend of daily orders. We can observe fluctuations and potential growth or decline over the recorded period.")
elif df is not None:
    print("Skipping daily orders plot: 'order_date' or 'order_id' column not found or dataframe is empty.")
```

```python
if df is not None and 'order_date' in df.columns and 'revenue' in df.columns:
    # Aggregate daily revenue
    daily_revenue = df.groupby('order_date')['revenue'].sum().reset_index()
    daily_revenue.columns = ['Date', 'Total Revenue']

    plt.figure(figsize=(12, 6))
    sns.lineplot(x='Date', y='Total Revenue', data=daily_revenue)
    plt.title('Daily Total Revenue Over Time')
    plt.xlabel('Date')
    plt.ylabel('Total Revenue')
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

    print("\nInterpretation: This line plot illustrates the daily revenue generated. Similar to orders, we can identify revenue peaks, dips, and overall trends.")
elif df is not None:
    print("Skipping daily revenue plot: 'order_date' or 'revenue' column not found or dataframe is empty.")
```
