# Restaurant Branch Performance Dataset - Exploratory Data Analysis (EDA)

This notebook will perform an in-depth Exploratory Data Analysis on the Restaurant Branch Performance Dataset to uncover relationships and patterns within the data.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configure display options for better readability
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

# Load the dataset
try:
    df = pd.read_csv('restaurant_branch_performance.csv')
    print("Dataset 'restaurant_branch_performance.csv' loaded successfully!")
except FileNotFoundError:
    print("Error: 'restaurant_branch_performance.csv' not found. Please upload the dataset or specify the correct path.")
    df = None # Set df to None if loading fails

if df is not None:
    # Display the first 5 rows of the dataset
    print("\nFirst 5 rows of the dataset:")
    display(df.head())

    # Display basic information about the dataset (columns, data types, non-null values)
    print("\nDataset Information:")
    df.info()

    # Display descriptive statistics for numerical columns
    print("\nDescriptive Statistics for Numerical Features:")
    display(df.describe())

    # Display descriptive statistics for categorical columns
    print("\nDescriptive Statistics for Categorical Features:")
    display(df.describe(include='object'))
```

## 1. Initial Data Cleaning and Preprocessing

Before proceeding with detailed analysis, we'll ensure data types are correct and handle any obvious inconsistencies.

```python
if df is not None:
    # Check for missing values
    print("\nMissing Values per Column:")
    display(df.isnull().sum().sort_values(ascending=False))

    # Convert any potential date columns to datetime objects if applicable
    # Assuming there might be a 'Date' or 'Order_Date' column. Adjust if actual column names differ.
    date_columns = [col for col in df.columns if 'date' in col.lower() or 'timestamp' in col.lower()]
    for col in date_columns:
        try:
            df[col] = pd.to_datetime(df[col])
            print(f"Column '{col}' converted to datetime.")
        except Exception as e:
            print(f"Could not convert '{col}' to datetime: {e}")

    print("\nUpdated Dataset Information after type conversion (if any):")
    df.info()
else:
    print("Cannot perform data cleaning as the DataFrame is not loaded.")
```

## 2. Distribution Analysis

Let's visualize the distributions of key numerical and categorical variables to understand their spread and unique values.

```python
if df is not None:
    # Distributions of numerical features (histograms)
    numerical_cols = df.select_dtypes(include=np.number).columns.tolist()
    print("\nVisualizing Distributions of Numerical Features:")

    plt.figure(figsize=(18, 12))
    for i, col in enumerate(numerical_cols):
        plt.subplot(np.ceil(len(numerical_cols)/3).astype(int), 3, i + 1)
        sns.histplot(df[col], kde=True, bins=20)
        plt.title(f'Distribution of {col}')
        plt.xlabel(col)
        plt.ylabel('Frequency')
    plt.tight_layout()
    plt.show()

    # Distributions of categorical features (count plots)
    categorical_cols = df.select_dtypes(include='object').columns.tolist()
    print("\nVisualizing Distributions of Categorical Features:")

    plt.figure(figsize=(18, len(categorical_cols)*4 if len(categorical_cols) > 0 else 6))
    for i, col in enumerate(categorical_cols):
        plt.subplot(len(categorical_cols), 2, i*2 + 1) # Arrange in 2 columns for better view
        sns.countplot(y=df[col], order=df[col].value_counts().index, palette='viridis')
        plt.title(f'Count of {col}')
        plt.xlabel('Count')
        plt.ylabel(col)
        plt.tight_layout()
    plt.show()
else:
    print("Cannot visualize distributions as the DataFrame is not loaded.")
```
